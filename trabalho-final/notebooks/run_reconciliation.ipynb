```python
import yaml

with open("../config/config.yaml") as f:
    config = yaml.safe_load(f)

for item in config["collections"]:
    tabela = f"{config['catalog']}.bronze.{item['collection']}"
    if not spark.catalog.tableExists(tabela):
        continue

    df = spark.table(tabela)
    total = df.count()
    nulos_chave = df.filter("_source_id IS NULL").count()
    duplicados = (df.groupBy("_source_id", "_ingestion_date").count()
                    .filter("count > 1").count())

    print(f"{item['collection']}: total={total} | _source_id nulo={nulos_chave} | duplicados={duplicados}")
```
